```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data_url = "https://raw.githubusercontent.com/amankharwal/Website-data/master/unemployment.csv"
df = pd.read_csv(data_url)

print(df.head())
print(df.info())

# Column naam ki jaisi bhi problem ho, hum seedha 3rd index wala column utha lenge
plt.figure(figsize=(12, 6))
sns.histplot(df.iloc[:, 3], bins=30, kde=True, color='purple')
plt.title("Unemployment Rate Distribution in India")
plt.xlabel("Unemployment Rate")
plt.ylabel("Count")
plt.show()

print("Task 2 Successfully Completed!")
```
